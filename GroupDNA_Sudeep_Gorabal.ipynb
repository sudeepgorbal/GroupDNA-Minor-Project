```python
parsed_messages = []
system_messages = 0
media_omitted = 0
deleted_messages = 0

with open('DADS Minor PROJECT dataset.txt', 'r', encoding='utf-8') as f:
    lines = f.readlines()

for line in lines:
    line = line.strip()
    if not line:
        continue

    try:
        timestamp_part, rest = line.split(' - ', 1)
    except ValueError:
        try:
            timestamp_part, rest = line.split(' ', 1)
        except ValueError:
            continue

    if ':' not in rest:
        system_messages += 1
        continue

    sender, text = rest.split(':', 1)
    sender = sender.strip()
    text = text.strip()

    if text == '<Media omitted>':
        media_omitted += 1
    elif text == 'This message was deleted':
        deleted_messages += 1
    else:
        parsed_messages.append({
            'timestamp': timestamp_part.strip(),
            'sender': sender,
            'text': text
        })

print(f"Successfully parsed {len(parsed_messages)} messages.")
print(f"Skipped {system_messages} system messages, {media_omitted} media-omitted, {deleted_messages} deleted messages.")

print("\nFirst 5 messages:")
for msg in parsed_messages[:5]:
    print(msg)

print("\nLast 5 messages:")
for msg in parsed_messages[-5:]:
    print(msg)
```

```python
from datetime import datetime

total_messages = len(parsed_messages)
first_timestamp = parsed_messages[0]['timestamp']
last_timestamp = parsed_messages[-1]['timestamp']

date_format = "%d/%m/%y, %H:%M"
start_date = datetime.strptime(first_timestamp, date_format)
end_date = datetime.strptime(last_timestamp, date_format)

total_days = (end_date - start_date).days + 1
period_str = f"{start_date.strftime('%d %B %Y')} to {end_date.strftime('%d %B %Y')} ({total_days} days)"

sender_counts = {}
for msg in parsed_messages:
    sender = msg['sender']
    sender_counts[sender] = sender_counts.get(sender, 0) + 1

sorted_senders = sorted(sender_counts.items(), key=lambda item: item[1], reverse=True)
num_participants = len(sender_counts)

print("========================================================")
print("GROUP OVERVIEW")
print("========================================================")
print(f"{'Group':<20}: DADS Minor PROJECT dataset")
print(f"{'Period':<20}: {period_str}")
print(f"{'Total messages':<20}: {total_messages:,}")
print(f"{'Participants':<20}: {num_participants}")
print("\nMESSAGES PER PERSON")
for sender, count in sorted_senders:
    pct = (count / total_messages) * 100
    print(f"{sender:<10}: {count:<5} ({pct:.1f}%)")
```

```python
import string
from datetime import datetime

date_counts = {}
hour_counts = {}

for msg in parsed_messages:
    dt = datetime.strptime(msg['timestamp'], "%d/%m/%y, %H:%M")
    date_str = dt.strftime('%d %B %Y')
    hour_str = dt.strftime('%H:00')

    date_counts[date_str] = date_counts.get(date_str, 0) + 1
    hour_counts[hour_str] = hour_counts.get(hour_str, 0) + 1

busiest_day = max(date_counts.items(), key=lambda x: x[1])
busiest_hour = max(hour_counts.items(), key=lambda x: x[1])

print(f"{'Busiest day':<17}: {busiest_day[0]} ({busiest_day[1]} messages)")
print(f"{'Busiest hour':<17}: {busiest_hour[0]}\n")

stop_words = {
    'i', 'is', 'the', 'a', 'and', 'or', 'to', 'of', 'in', 'on', 'for', 'it', 'my', 'me',
    'you', 'your', 'that', 'this', 'hai', 'ki', 'se', 'ko', 'bhi', 'na', 'toh', 'ka', 'ke',
    'ho', 'are', 'not', 'do', 'dont', 'be', 'was', 'have', 'but', 'as', 'at', 'so', 'we',
    'they', 'with', 'what', 'if', 'all', 'can', 'will', 'just', 'like', 'how', 'ye', 'woh'
}

word_counts = {}
for msg in parsed_messages:
    text = msg['text'].lower()
    for p in string.punctuation + "“”‘’":
        text = text.replace(p, '')

    for word in text.split():
        if word not in stop_words and len(word) > 1:
            word_counts[word] = word_counts.get(word, 0) + 1

sorted_words = sorted(word_counts.items(), key=lambda item: item[1], reverse=True)[:10]

print("THIS GROUP'S FAVOURITE WORDS")
if sorted_words:
    max_count = sorted_words[0][1]
    for word, count in sorted_words:
        bar_length = max(1, int((count / max_count) * 20))
        print(f"{word:<10} {'█' * bar_length} {count}")
```

```python
import numpy as np

participants = [p[0] for p in sorted_senders[:6]]
participant_idx = {p: i for i, p in enumerate(participants)}

heatmap = np.zeros((len(participants), 24), dtype=int)

for msg in parsed_messages:
    sender = msg['sender']
    if sender in participant_idx:
        hour = int(msg['timestamp'].split(', ')[1].split(':')[0])
        heatmap[participant_idx[sender], hour] += 1

print("ACTIVITY HEATMAP (hour of day, columns 00 to 23)")
print("          00 03 06 09 12 15 18 21")

for i, person in enumerate(participants):
    row = heatmap[i]
    max_val = row.max() if row.max() > 0 else 1

    row_str = ""
    for j, val in enumerate(row):
        pct = val / max_val
        if pct == 0:
            char = " "
        elif pct <= 0.25:
            char = "."
        elif pct <= 0.50:
            char = "░"
        elif pct <= 0.75:
            char = "▒"
        else:
            char = "▓"

        row_str += char
        if j in [2, 5, 8, 11, 14, 17, 20]:
            row_str += " "

    print(f"{person:<10}{row_str}")
```

```python
from datetime import datetime, timedelta

# Compute response times and streaks
response_times = {p[0]: [] for p in sorted_senders}
last_msg_time = None
last_sender = None

# Track days active per person for silent streaks
person_days = {p[0]: set() for p in sorted_senders}
all_chat_dates = sorted(list(date_counts.keys()))

start_dt = datetime.strptime(all_chat_dates[0], '%d %B %Y') if all_chat_dates else datetime.now()
end_dt = datetime.strptime(all_chat_dates[-1], '%d %B %Y') if all_chat_dates else datetime.now()

for msg in parsed_messages:
    dt = datetime.strptime(msg['timestamp'], "%d/%m/%y, %H:%M")
    sender = msg['sender']
    date_str = dt.strftime('%d %B %Y')

    if sender in person_days:
        person_days[sender].add(date_str)

    if last_sender and last_sender != sender:
        gap_seconds = (dt - last_msg_time).total_seconds()
        if 0 <= gap_seconds < 86400: # filter out massive gaps for "response time"
            response_times[sender].append(gap_seconds)

    last_msg_time = dt
    last_sender = sender

avg_response_times = {}
for person, times in response_times.items():
    if times:
        avg_mins = sum(times) / len(times) / 60
        avg_response_times[person] = avg_mins
    else:
        avg_response_times[person] = 999999

sorted_repliers = sorted(avg_response_times.items(), key=lambda x: x[1])

print("RESPONSE PATTERNS")
if sorted_repliers:
    fastest = sorted_repliers[0]
    slowest = sorted_repliers[-1]

    fast_str = f"{int(fastest[1])} minutes" if fastest[1] < 120 else f"{fastest[1]/60:.1f} hours"
    slow_str = f"{slowest[1]/60:.1f} hours" if slowest[1] >= 120 else f"{int(slowest[1])} minutes"

    print(f"Fastest replier : {fastest[0]} (avg {fast_str})")
    print(f"Slowest replier : {slowest[0]} (avg {slow_str})\n")

print("LONGEST SILENT STREAKS")
all_dates_list = []
curr = start_dt
while curr <= end_dt:
    all_dates_list.append(curr.strftime('%d %B %Y'))
    curr += timedelta(days=1)

for person in participants:
    active_set = person_days[person]
    max_streak = 0
    current_streak = 0

    for d in all_dates_list:
        if d not in active_set:
            current_streak += 1
            if current_streak > max_streak:
                max_streak = current_streak
        else:
            current_streak = 0

    print(f"{person:<10} : {max_streak} days silent")
```

```python
# Personality Archetype Detection
print("PERSONALITY ARCHETYPES\n")

# Calculate metrics per person for archetype rules
for person in participants:
    p_msgs = [m for m in parsed_messages if m['sender'] == person]
    total_p_msgs = len(p_msgs)

    if total_p_msgs == 0:
        continue

    # 1. Night Owl check (msgs between 23:00 and 04:59)
    night_msgs = 0
    for m in p_msgs:
        hour = int(m['timestamp'].split(', ')[1].split(':')[0])
        if hour >= 23 or hour <= 4:
            night_msgs += 1
    night_pct = (night_msgs / total_p_msgs) * 100

    # 2. Storyteller check (avg words per message)
    total_words = sum(len(m['text'].split()) for m in p_msgs)
    avg_words = total_words / total_p_msgs

    # 3. Drama Queen check (all-caps messages)
    all_caps_count = sum(1 for m in p_msgs if m['text'].isupper() and len(m['text']) > 2)
    caps_pct = (all_caps_count / total_p_msgs) * 100

    # 4. Ghost check (silent days percentage)
    active_days_count = len(person_days[person])
    silent_days_pct = ((total_days - active_days_count) / total_days) * 100

    # Assign archetype based on highest matching threshold rule
    archetype = "THE REGULAR"
    detail = ""

    if night_pct > 60:
        archetype = "THE NIGHT OWL"
        detail = f"({night_pct:.1f}% msgs after 11 PM)"
    elif avg_words > 25:
        archetype = "THE STORYTELLER"
        detail = f"(avg {avg_words:.1f} words per msg)"
    elif caps_pct > 25:
        archetype = "THE DRAMA QUEEN"
        detail = f"({caps_pct:.1f}% ALL-CAPS msgs)"
    elif silent_days_pct > 60:
        archetype = "THE GHOST"
        detail = f"(silent {silent_days_pct:.0f}% of days)"
    elif total_p_msgs > 800:
        archetype = "THE SPAMMER"
        detail = f"(high volume: {total_p_msgs} msgs)"
    else:
        archetype = "THE GROUP MOM"
        detail = "(caring/supportive contributor)"

    print(f"{person:<10} -> {archetype:<18} {detail}")
```
